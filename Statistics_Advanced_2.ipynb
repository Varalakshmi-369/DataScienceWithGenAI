**Question 1:**

What is hypothesis testing in statistics?

**Answer:**

Hypothesis testing is a way to decide whether there’s enough evidence in a sample to support a claim about a population.

**Example:**

A company claims that their bulbs last 1000 hours.
You test 50 bulbs → average = 980 hours.
You perform hypothesis testing to check if the difference (980 vs. 1000) is just random or truly significant.

**Question 2:**

What is the null hypothesis, and how does it differ from the alternative
hypothesis?

**Answer:**

**Null hypothesis (H₀):** Assumes no effect or no difference — the default assumption.

**Alternative hypothesis (H₁ or Hₐ):** Claims that there is an effect or difference.

**Example:**

H₀: Mean weight = 50 kg

H₁: Mean weight ≠ 50 kg


You collect data to test if you should reject H₀ in favor of H₁.

**In short:**

H₀ = “Nothing’s changed.”

H₁ = “Something’s different.”

**Question 3:**

 Explain the significance level in hypothesis testing and its role in deciding
the outcome of a test.

**Answer:**

The significance level (α) is the cutoff for deciding when to reject H₀.
Usually, α = 0.05 (5%).

**Meaning:**

If the probability (p-value) of seeing your data, assuming H₀ is true, is less than 0.05, it’s considered too rare → you reject H₀.

**In short:**

α = Risk you’re willing to take of being wrong when rejecting H₀.

**Question 4:**

What are Type I and Type II errors? Give examples of each.

**Answer:**

- **Type I error (False Positive):** Rejecting H₀ when it is actually true. Probability = α. Example: Concluding a new drug works when it actually doesn't.
- **Type II error (False Negative):** Failing to reject H₀ when H₁ is true. Probability = β. Example: Concluding the drug has no effect when it actually does.

There's a trade-off: lowering α (less Type I errors) usually increases β (more Type II errors), unless you increase sample size.

**In short:**

Type I = False alarm

Type II = Missed detection


**Question 5:**

What is the difference between a Z-test and a T-test? Explain when to use
each.

**Answer:**

- **Z-test:**
Used when the population standard deviation (σ) is known or when the sample size is large (so the sample standard deviation is a good approximation).It uses the normal (Z) distribution.

- **T-test:**
Used when σ is unknown and the sample size is small (typically n < 30). It uses Student's t-distribution, which has heavier tails to account for extra uncertainty.

**In practice:**

use a t-test when σ is unknown (most real problems).

Use a z-test when σ is known or n is large.



**Question 6:**

Write a Python program to generate a binomial distribution with n=10 and
p=0.5, then plot its histogram.

**Answer**


A binomial distribution describes the number of successes in a fixed number of independent trials, each with the same probability of success (p).

**For example:**

If you flip a coin 10 times (n=10) and each flip has a 50% chance of getting heads (p=0.5),
the distribution of how many heads you get follows a binomial distribution.

**Key Terms:**

n = 10: number of trials

p = 0.5: probability of success per trial

size = 10000: number of simulated experiments

np.random.binomial(): function to generate binomial random values

```python
# Question 6: Write a Python program to generate a binomial distribution with n=10 and p=0.5, then plot its histogram.

import numpy as np
import matplotlib.pyplot as plt

# Parameters
n = 10        # number of trials
p = 0.5       # probability of success
Nsim = 10000  # number of experiments

# Generate random binomial data
data = np.random.binomial(n, p, size=Nsim)

# Plot histogram
plt.figure(figsize=(6,4))
plt.hist(data, bins=range(n+2), edgecolor='black')
plt.title('Binomial Distribution (n=10, p=0.5)')
plt.xlabel('Number of successes')
plt.ylabel('Frequency')
plt.show()

# Print summary statistics
print("Mean of simulated data:", data.mean())
print("Variance of simulated data:", data.var())
```

Implement hypothesis testing using Z-statistics for a sample dataset in

Show the Python code and interpret the results.

50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

**Answer:**

Hypothesis testing helps decide whether the sample mean is significantly different from a population mean.


We’ll use a Z-test, which is appropriate when:

The population standard deviation (σ) is known, OR

The sample size is large (n ≥ 30) — in our case, n = 36.


sample_data = [
    49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
    50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
    50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
    50.3, 50.4, 50.0, 49.7, 50.5, 49.9
]

Sample size (n) = 36

Population mean (μ₀) = 50

We’ll assume σ = 1.0 (since not provided).

**Step 1:** Define Hypotheses

Null hypothesis (H₀): μ = 50

Alternative hypothesis (H₁): μ ≠ 50 (two-tailed test)

Step 2: Formula


```python
import numpy as np
from math import sqrt
from scipy import stats

# Given data
sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
               50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
               50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
               50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

# Convert to numpy array
data = np.array(sample_data)

# Step 1: Compute sample statistics
n = len(data)
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)  # sample std deviation

# Assume population standard deviation
sigma = 1.0

# Step 2: Compute Z-statistic
z_stat = (sample_mean - 50) / (sigma / sqrt(n))

# Step 3: Compute p-value (two-tailed)
p_value = 2 * (1 - stats.norm.cdf(abs(z_stat)))

print("Sample Size (n):", n)
print("Sample Mean:", round(sample_mean, 3))
print("Assumed Sigma:", sigma)
print("Z-Statistic:", round(z_stat, 3))
print("P-Value:", round(p_value, 6))
```

**Interpretation**

Significance Level (α) = 0.05

p-value = 0.548 > 0.05

Therefore, we fail to reject H₀.

There is no significant difference between the sample mean and the population mean (50).

In simple terms — the data supports that the average is roughly 50.

**Question 8:**

Write a Python script to simulate data from a normal distribution and
calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.

**Answer:**





A confidence interval (CI) gives a range of plausible values for a population parameter (like the mean) based on sample data.

For a 95% CI, it means:

“We are 95% confident that the true population mean lies within this range.”

```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Step 1: Simulate data
np.random.seed(42)   # for reproducibility
mu = 5.0             # true mean
sigma = 2.0          # true standard deviation
n = 100              # sample size

data = np.random.normal(loc=mu, scale=sigma, size=n)

# Step 2: Compute sample statistics
sample_mean = np.mean(data)
sample_std = np.std(data, ddof=1)

# Step 3: Find t-critical value for 95% CI
alpha = 0.05
df = n - 1
t_crit = stats.t.ppf(1 - alpha/2, df)

# Step 4: Compute confidence interval
se = sample_std / np.sqrt(n)
ci_lower = sample_mean - t_crit * se
ci_upper = sample_mean + t_crit * se

# Step 5: Print and visualize
print(f"Sample mean = {sample_mean:.3f}")
print(f"95% Confidence Interval = [{ci_lower:.3f}, {ci_upper:.3f}]")

# Step 6: Plot histogram
plt.figure(figsize=(6,4))
plt.hist(data, bins=15, edgecolor='black', color='skyblue')
plt.title('Simulated Normal Data (n=100)')
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.axvline(sample_mean, color='red', linestyle='--', label='Sample Mean')
plt.axvline(ci_lower, color='green', linestyle='--', label='95% CI lower')
plt.axvline(ci_upper, color='green', linestyle='--', label='95% CI upper')
plt.legend()
plt.tight_layout()
plt.show()
```

**Interpretation**


If you repeated the sampling many times, about 95% of such intervals would contain the true mean (μ = 5.0).


So, you can be 95% confident that the population mean lies within [4.49, 5.28] for this sample.

**Question 9:**

Write a Python function to calculate the Z-scores from a dataset and
visualize the standardized data using a histogram. Explain what the Z-scores represent
in terms of standard deviations from the mean.

**Answer:**

A Z-score tells you how far a data point is from the mean, measured in standard deviations.

It helps standardize data, meaning it converts all values to a common scale — so you can easily compare them, even if they originally had different units or scales.

| Z-score  | Meaning                                  |
| -------- | ---------------------------------------- |
| 0        | The value is exactly at the mean         |
| +1       | 1 standard deviation above the mean      |
| -1       | 1 standard deviation below the mean      |
| +2 / -2  | 2 SDs away (unusual but possible)        |
| +3 or -3 | Extremely far from mean (rare, <1% data) |



```python
import numpy as np
import matplotlib.pyplot as plt

# Step 1: Define a function to calculate Z-scores
def z_scores(data):
    data = np.array(data)
    mean = np.mean(data)
    std = np.std(data, ddof=0)  # population std
    z = (data - mean) / std
    return z

# Step 2: Example dataset
data = [10, 12, 9, 11, 13, 12, 8, 10, 11, 14]

# Step 3: Compute Z-scores
zs = z_scores(data)

# Step 4: Display results
print("Original Data:", data)
print("Mean:", round(np.mean(data), 2))
print("Standard Deviation:", round(np.std(data, ddof=0), 2))
print("Z-scores:", np.round(zs, 2))

# Step 5: Plot histogram of Z-scores
plt.figure(figsize=(6,4))
plt.hist(zs, bins=8, edgecolor='black', color='lightblue')
plt.title("Histogram of Z-scores (Standardized Data)")
plt.xlabel("Z-score (Standard Deviations from Mean)")
plt.ylabel("Frequency")
plt.axvline(0, color='red', linestyle='--', label='Mean (Z=0)')
plt.legend()
plt.tight_layout()
plt.show()
```

**Interpretation:**

Z-scores show how many standard deviations a data point is from the mean.

**For example:**

Z = 1.73 → value is 1.73 SD above the mean

Z = -1.73 → 1.73 SD below the mean

Helps detect outliers (values beyond ±3 are unusual).
