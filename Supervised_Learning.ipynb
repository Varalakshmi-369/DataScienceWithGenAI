### **Supervised Learning: Regression Models and Performance Metrics**

**Question 1: What is Simple Linear Regression (SLR)? Explain its purpose.**

**Answer:**

Simple Linear Regression (SLR) is a statistical technique that models the relationship between two continuous variables — one independent variable (X) and one dependent variable (Y).

It fits a straight line (best-fit line) through the data points that best predicts the value of Y based on X.

**Purpose:**

* To predict the dependent variable based on the independent variable.

* To understand how a change in X affects Y.

* To find the trend or relationship between two variables.

**Example:** Predicting a student’s marks (Y) based on the number of study hours (X).


**Question 2: What are the key assumptions of Simple Linear Regression?**

**Answer:**

For the SLR model to give accurate predictions, the following assumptions must hold:

**Linearity**: The relationship between X and Y is linear.

**Independence**: Observations are independent of each other.

**Homoscedasticity**: Constant variance of residuals (errors) across all values of X.

**Normality**: The residuals (errors) are normally distributed.

**No multicollinearity:** (Not relevant for SLR, only for multiple regression).



**Why it matters for development:**

Violating these assumptions can make your model predictions biased or inaccurate, so always visualize data (scatter plots, residual plots) before fitting.

**Question 3: Write the mathematical equation for a simple linear regression model and explain each term.**

**Answer:**


The mathematical equation for **Simple Linear Regression** is:

### Simple Linear Regression Equation

$$
Y = \beta_0 + \beta_1 X + \varepsilon
$$

**Where:**

- \(Y\): Dependent variable  (what we predict)

- \(X\): Independent variable  (input feature)

- \(β0): Intercept  (value of Y when X = 0)

- \(β1): Slope (change in Y for one unit change in X)  

- \(ε): Error term  (difference between actual and predicted Y)

---



### **Interpretation**

If \( β1 = 5 ), then for every 1-unit increase in \( X \), \( Y \) increases by **5 units** on average.

**Question 5: What is the method of least squares in linear regression?**

**Answer:**


The method of least squares finds the line that minimizes the sum of squared errors between the actual and predicted values.

### Method of Least Squares (Linear Regression)

The goal is to find the best-fit line by minimizing the sum of squared errors:

$$
\text{Minimize } \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2
$$

**Where:**


* $Y_i  = actual value$
* $\hat{Y}_i  = predicted value from the model $

 **Why Squared?**

Squaring ensures errors are positive and penalizes large deviations more heavily:

$$
(Y_i - \hat{Y}_i)^2
$$

 **Developer Insight:**

Under the hood, scikit-learn’s **LinearRegression()** uses the **Least Squares Method** to estimate:

$$
\beta_0 \quad \text{and} \quad \beta_1
$$



**Question 6: What is Logistic Regression? How does it differ from Linear Regression?**

**Answer:**

Logistic Regression is used for classification, not regression.
It predicts the probability that an observation belongs to a certain class (e.g., spam vs. not spam).

**Linear Regression vs Logistic Regression:**


| Feature     | Linear Regression        | Logistic Regression                   |
| ----------- | ------------------------ | ------------------------------------- |
| Output Type | Continuous (real values) | Categorical (probabilities 0–1)       |
| Equation    | (Y = β_0 + β_1X)         | (P = \frac{1}{1 + e^{-(β_0 + β_1X)}}) |
| Use Case    | Predict prices, scores   | Predict binary outcomes               |
| Example     | Predict salary           | Predict if email is spam              |


**Question 7: Name and briefly describe three common evaluation metrics for regression models.**


 **Regression Evaluation Metrics – Formulas**

 **1. Mean Absolute Error (MAE)**

 * Average of absolute differences between actual and predicted values.

$$
MAE = \frac{1}{n} \sum_{i=1}^{n} \left| Y_i - \hat{Y}_i \right|
$$

→ Easy to interpret in same units as Y.

 **2. Mean Squared Error (MSE)**
 * Average of squared differences.

$$
MSE = \frac{1}{n} \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2
$$

→ Penalizes larger errors more.

 **3. Root Mean Squared Error (RMSE)**
* Square root of MSE — gives error in same units as Y.
$$
RMSE = \sqrt{ \frac{1}{n} \sum_{i=1}^{n} (Y_i - \hat{Y}_i)^2 }
$$

Use in development:

MAE → good for interpretability,

RMSE → sensitive to outliers.

 **Where:**

* $ Y_i  = actual value  $
* $\hat{Y}_i = predicted value  $
* $ n  = number of samples  $


**Question 8: What is the purpose of the R-squared metric in regression analysis?**

**Answer:**


R-squared (R²) measures how well the regression line fits the data.
It represents the proportion of the variance in Y that is explained by X.

**R-Squared (Coefficient of Determination)**

The formula for R-squared is:

$$
R^2 = 1 - \frac{SS_{res}}{SS_{tot}}
$$


 **Meaning:**

* $ R^2 = 1  → Perfect model  $
* $ R^2 = 0  → Model explains nothing  $


**Developer note:**

Higher R² → better model, but be careful: overfitting can artificially increase R² in multiple regression.

**Question 9: Write Python code to fit a simple linear regression model using scikit-learn and print the slope and intercept.**

**Answer:**

```python
# Import libraries
import numpy as np
from sklearn.linear_model import LinearRegression

# Example data
X = np.array([1, 2, 3, 4, 5]).reshape(-1, 1)  # Independent variable
y = np.array([2, 4, 5, 4, 5])                 # Dependent variable

# Create and fit model
model = LinearRegression()
model.fit(X, y)

# Print coefficients
print("Slope (β1):", model.coef_[0])
print("Intercept (β0):", model.intercept_)

# Predict value
pred = model.predict([[6]])
print("Predicted value for X=6:", pred[0])
```

**Question 10: How do you interpret the coefficients in a simple linear regression model?**

**Answer:**

**Intercept (β₀):** The predicted value of Y when X = 0.

**Slope (β₁):** The average change in Y for a one-unit change in X.

**Example:**
If Y = 3 + 2X,

→ When X increases by 1, Y increases by 2.

→ When X = 0, predicted Y = 3.

**In practice:**

If you’re predicting sales based on ad spend, a slope of 50 means for every extra \$1 spent on ads, sales increase by \$50.
